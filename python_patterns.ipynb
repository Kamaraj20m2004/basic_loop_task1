**2.print the below pattern**

```python
#
##
###
####
#####

i=1
while i<=5:
    j=1
    while j<=i:
        print("#",end="")
        j+=1
    i+=1
    print()
```

```python
for i in range(5):
    for j in range(i+1):
        print("",end="")

    for k in range(5-i):
        print("&",end="")
    print()
```
